```python
import pandas as pd
```

```python
carreras = pd.read_csv('carreras_cursos_up.csv')
habilidades = pd.read_csv('up_sumillas_habilidades.csv')

# Limpiar nombre de carrera y filtrar solo carreras de Ingeniería (+ Humanidades Digitales)
carreras['carrera'] = (
    carreras['carrera']
    .str.replace('Plan de Estudios | ', '', regex=False)
    .str.replace('Carrera de ', '', regex=False)
    .str.replace(' | UP', '', regex=False)
    .str.strip()
)

carreras_inge = [
    'Ingeniería en Innovación y Diseño',
    'Ingeniería Empresarial',
    'Ingeniería de la Información',
    'Humanidades Digitales',
]
carreras = carreras[carreras['carrera'].isin(carreras_inge)].copy()
carreras['curso'] = carreras['curso'].str.replace('\u200b', '', regex=False).str.replace('\xa0', ' ', regex=False).str.strip()
```

```python
carreras_habilidades = pd.merge(carreras, habilidades, on='curso', how='left')
```
